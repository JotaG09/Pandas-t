```python
# As chaves são as colunas e os valores são as linhas, ou seja, os dados de cada pessoa.

people = {
    "first": ["Corey", 'Jane', 'John'],
    "last": ["Schafer", 'Doe', 'Doe'],
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['first'] + " " + df['last']
```

```python
df['nome_completo'] = df['first'] + " " + df['last']
```

```python
df
```

USANDO MÉTODO .drop()

```python
df.drop(columns = ['first', 'last'], inplace = True)
# O método drop é usado para remover as colunas 'first' e 'last' do DataFrame df. O parâmetro columns especifica quais colunas devem ser removidas,
# e o parâmetro inplace=True indica que a operação deve ser realizada no próprio DataFrame
```

```python
df
```

```python
df['nome_completo'].str.split(' ', expand = True)
# O método str.split é usado para dividir os valores da coluna 'nome_completo' em duas partes, usando o espaço como delimitador.
# O parâmetro expand=True indica que o resultado deve ser expandido em colunas separadas, resultando em um DataFrame com duas colunas,
# uma para o primeiro nome e outra para o sobrenome.
```

```python
df
```

```python
df[['first', 'last']] = df['nome_completo'].str.split(' ', expand = True)
# O código df[['first', 'last']] = df['nome_completo'].str.split(' ', expand=True) divide a coluna 'nome_completo' em duas partes (first e last)
# O parâmetro expand = True garante que o resultado seja expandido em colunas separadas.
```
