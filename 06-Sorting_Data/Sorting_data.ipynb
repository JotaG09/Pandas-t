```python
# As chaves são as colunas e os valores são as linhas, ou seja, os dados de cada pessoa.

people = {
    "first": ["Corey", 'Jane', 'John','Adão'],
    "last": ["Schafer", 'Doe', 'Doe','Silva'],
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com','AdaoSilva@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.sort_values(by='last') # Ordena o DataFrame com base na coluna 'last' (sobrenome) em ordem alfabética.
```

```python
df.sort_values(by='last', ascending=False) # Ordena o DataFrame com base na coluna 'last' (sobrenome) em ordem alfabética decrescente.
```

```python
df.sort_values(by=['last', 'first']) # Ordena o DataFrame primeiro pela coluna 'last' e, em caso de empate, pela coluna 'first'.
# Exemplo: Se houver duas pessoas com o mesmo sobrenome, elas serão ordenadas pelo nome.
```

```python
df.sort_values(by=['last', 'first'], ascending=[False, True])
# Ordena o DataFrame primeiro pela coluna 'last' em ordem alfabética decrescente e, em caso de empate, pela coluna 'first' em ordem alfabética crescente.
```

```python
df.sort_index()
```

```python
df['last'].sort_values() # Ordena apenas a coluna 'last' em ordem alfabética, retornando uma Series ordenada.
```
