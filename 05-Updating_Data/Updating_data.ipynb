```python
people = {
    "first": ["Corey", 'Jane', 'John'],
    "last": ["Schafer", 'Doe', 'Doe'],
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.columns
```

```python
df.columns = ['first_name','last_name', 'email']
```

```python
df
```

```python
df.columns = [x.upper() for x in df.columns]
df
```

```python
df.columns = df.columns.str.replace('_',' ')
df
```

```python
df.columns = df.columns.str.replace(' ','_')
df.columns = [x.lower() for x in df.columns] # Converte os nomes das colunas para letras minúsculas usando uma list comprehension.
```

```python
df
```

```python
df.rename(columns={'first_name':'first', 'last_name': 'last'}, inplace = True)
# Renomeia as colunas 'first_name' para 'first' e 'last_name' para 'last' usando o método rename do DataFrame.
df
```

```python
df.loc[2] = ['John','Smith','JhonSmith@email.com']
df
```

```python
df.loc[2, ['last','email']]
```

```python
df.loc[2, ['last','email']] = ['Matheus', 'Matheus@email.com']
df
```

```python
df.loc[2,'last'] = 'Carlos'
df
```

```python

# Usando o método at para atualizar um valor específico
# O método at é mais rápido do que loc para acessar um único valor, pois é otimizado para esse tipo de operação.
df.at[2,'last'] = 'Matheus'
df
```

```python
filtro = (df['email'] == 'JaneDoe@email.com')
df.loc[filtro, 'last'] = 'Smith'
df

# É necessário usar o método loc para atualizar os valores
# pois ele permite acessar e modificar os valores de forma eficiente, especialmente quando se trata de condições ou filtros.
# O método at é mais adequado para acessar um único valor específico.
```

```python
filtro = (df['email'] == 'JaneDoe@email.com')
df.loc[filtro, 'last'] = 'Smith'
df
```

USANDO A FUNÇÃO .apply()

```python
df['email'].apply(len) # Aplica a função len a cada valor da coluna 'email', retornando o comprimento de cada string de email.
```

```python
def update_email(email):
    return email.upper()  # ou email.upper() para converter para maiúsculas
```

```python
df['email'].apply(update_email) # Aplica a função update_email a cada valor da coluna 'email', convertendo os emails para letras maiúsculas.
df
```

```python
df['email'] = df['email'].apply(update_email) # Atualiza a coluna 'email' com os valores convertidos para letras minúsculas usando a função update_email.
df
```

```python
df.apply(len, axis = 'columns') # Aplica a função len a cada linha do DataFrame, retornando o número de colunas em cada linha.
```

```python
df.apply(len, axis='rows') # Aplica a função len a cada coluna do DataFrame, retornando o número de linhas em cada coluna.
```

USANDO A FUNÇÃO .map()

```python
df['first'].map({'Corey': 'João', 'Jane': 'Janete'}) # Substitui os valores da coluna 'first' com base em um dicionário de mapeamento;
# onde as chaves são os valores originais e os valores são os novos valores a serem atribuídos.
# não alteram o DataFrame original, a menos que sejam atribuídos de volta à coluna ou ao DataFrame.
```

```python
df
```

```python
df['first'].replace({'Corey': 'João', 'Jane': 'Janete'})
# o método replace é usado para substituir os valores da coluna 'first' com base em um dicionário de mapeamento,
# onde as chaves são os valores originais e os valores são os novos valores a serem atribuídos.
```

```python
df['first'] = df['first'].replace({'Corey': 'João', 'Jane': 'Janete'})
# O DataFrame df é atualizado com os novos valores na coluna 'first', onde 'Corey' é substituído por 'João' e 'Jane' é substituído por 'Janete'.
```
